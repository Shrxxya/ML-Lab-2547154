```python
import pandas as pd

from sklearn.preprocessing import OrdinalEncoder
from sklearn.model_selection import train_test_split

from sklearn.naive_bayes import CategoricalNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report
```

```python
df = pd.read_csv("Lab 8 - Sheet1.csv")

print("Dataset")
df.head()
```

```python
X = df[['Outlook','Temperature','Humidity','Wind']]

y = df['Play Tennis']

print("\nFeatures")
print(X)

print("\nTarget")
print(y)
```

The model will use the weather conditions to predict whether a person will play tennis.

```python
encoder = OrdinalEncoder()

X_encoded = encoder.fit_transform(X)

print("\nEncoded Features")
print(X_encoded[:3])
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded,
    y,
    test_size=0.3,
    random_state=42,
    stratify=y
)

print("\nTraining Samples:", len(X_train))
print("Testing Samples:", len(X_test))
```

The dataset is divided into training and testing sets using a 70:30 ratio, and stratified so as to ensure equal representation of each class.

Categorical Naive Bayes Classifier

```python
nb_model = CategoricalNB()

nb_model.fit(X_train, y_train)
```

```python
y_pred = nb_model.predict(X_test)

print("\nPredicted Labels")
print(y_pred)
```

```python
accuracy = accuracy_score(y_test, y_pred)

print("\nNaive Bayes Accuracy =", accuracy)
```

The Categorical Naive Bayes model achieved an accuracy of 86.67%, meaning it correctly classified approximately 87 out of every 100 test samples.

```python
cm = confusion_matrix(y_test, y_pred)

print("\nConfusion Matrix")
print(cm)
```

There were only 2 misclassifications out of 15

```python
report = classification_report(y_test, y_pred)

print("\nClassification Report")
print(report)
```

```python
new_data = pd.DataFrame({
    'Outlook':['Sunny'],
    'Temperature':['Cool'],
    'Humidity':['High'],
    'Wind':['Strong']
})

new_encoded = encoder.transform(new_data)

prediction = nb_model.predict(new_encoded)

print("\nPrediction:")
print(prediction[0])
```

For the given weather condition (Sunny, Cool, High Humidity, Strong Wind), the model predicts that the person will not play tennis, as the conditions were deemed unfavourable.

```python
probability = nb_model.predict_proba(new_encoded)

print("\nPrediction Probability")

print("No  =", probability[0][0])
print("Yes =", probability[0][1])
```

The model assigns a probability of 74.66% to the "No" class and 25.34% to the "Yes" class. Since the probability of "No" is higher, the final prediction is No.

Decision Tree Model

```python
dt_model = DecisionTreeClassifier(random_state=42)

dt_model.fit(X_train, y_train)

dt_pred = dt_model.predict(X_test)

dt_accuracy = accuracy_score(y_test, dt_pred)

print("\nDecision Tree Accuracy =", dt_accuracy)
```

The Decision Tree classifier achieved 100% accuracy, correctly classifying all test samples.

```python
dt_prediction = dt_model.predict(new_encoded)

print("\nDecision Tree Prediction:")
print(dt_prediction[0])
dt_probability = dt_model.predict_proba(new_encoded)

print("\nDecision Tree Probability")

for cls, prob in zip(dt_model.classes_, dt_probability[0]):
    print(cls, ":", prob)
```

Logistic Regression Model

```python
lr_model = LogisticRegression(max_iter=200)

lr_model.fit(X_train, y_train)

lr_pred = lr_model.predict(X_test)

lr_accuracy = accuracy_score(y_test, lr_pred)

print("\nLogistic Regression Accuracy =", lr_accuracy)
```

The Logistic Regression classifier achieved 93.33% accuracy, indicating excellent predictive performance with only a small number of incorrect classifications.

```python
lr_prediction = lr_model.predict(new_encoded)

print("\nLogistic Regression Prediction:")
print(lr_prediction[0])
lr_probability = lr_model.predict_proba(new_encoded)

print("\nLogistic Regression Probability")

for cls, prob in zip(lr_model.classes_, lr_probability[0]):
    print(cls, ":", prob)
```

```python
print("\nModel Comparison")

print("Categorical Naive Bayes :", accuracy)

print("Decision Tree           :", dt_accuracy)

print("Logistic Regression     :", lr_accuracy)
```

Since the dataset is relatively small, the perfect accuracy may also indicate that the Decision Tree has learned the specific patterns in this dataset very closely.

Logistic Regression achieved an accuracy of 93.33%, which indicates that it was able to effectively learn the relationship between weather conditions and the decision to play tennis.

The slightly lower accuracy of Naive Bayes may be because it assumes that the input features are independent of each other, while in real-world datasets some features such as temperature, humidity, and weather conditions may have relationships.
